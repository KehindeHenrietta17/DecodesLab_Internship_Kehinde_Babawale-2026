```python
conn = sqlite3.connect("project3_sales.db")
```

```python
df.to_sql("orders", conn, if_exists="replace", index=False)
```

```python
query = """
SELECT name
FROM sqlite_master
WHERE type = 'table';
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT *
FROM orders;
"""

result = pd.read_sql_query(query, conn)

result.head()
```

```python
query = """
SELECT OrderID, Product, Quantity, TotalPrice
FROM orders;
"""

result = pd.read_sql_query(query, conn)

result.head(10)
```

```python
query = """
SELECT OrderID, Product, Quantity, TotalPrice
FROM orders
WHERE TotalPrice > 2000;
"""

result = pd.read_sql_query(query, conn)

result.head(10)
```

```python
query = """
SELECT OrderID, Product, Quantity, TotalPrice
FROM orders
WHERE Quantity > 4;
"""

result = pd.read_sql_query(query, conn)

result.head(10)
```

```python
query = """
SELECT OrderID, Product, Quantity, TotalPrice
FROM orders
ORDER BY TotalPrice DESC;
"""

result = pd.read_sql_query(query, conn)

result.head(10)
```

```python
query = """
SELECT 
    Product,
    COUNT(*) AS Number_of_Orders
FROM orders
GROUP BY Product
ORDER BY Number_of_Orders DESC;
"""

result = pd.read_sql_query(query, conn)

result
```

#### Finding

**Printer** had the highest number of orders with **181**, while **Phone** had the lowest with **156**.

```python
query = """
SELECT 
    Product,
    SUM(TotalPrice) AS Total_Order_Value
FROM orders
GROUP BY Product
ORDER BY Total_Order_Value DESC;
"""

result = pd.read_sql_query(query, conn)

result
```

#### Finding

**Chair** generated the highest recorded order value at **195,620.11**, closely followed by Printer at **195,612.61**

```python
query = """
SELECT 
    Product,
    AVG(TotalPrice) AS Average_Order_Value
FROM orders
GROUP BY Product
ORDER BY Average_Order_Value DESC;
"""

result = pd.read_sql_query(query, conn)

result
```

#### Finding

**Laptop** had the highest average order value at approximately **1,110.56**.

```python
query = """
SELECT 
    Product,
    COUNT(*) AS Purchase_Frequency,
    SUM(TotalPrice) AS Total_Order_Value,
    AVG(TotalPrice) AS Avg_Transaction_Value
FROM orders
GROUP BY Product
ORDER BY Total_Order_Value DESC;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT 
    CustomerID,
    COUNT(*) AS Number_of_Orders
FROM orders
GROUP BY CustomerID
HAVING COUNT(*) > 1
ORDER BY Number_of_Orders DESC;
"""

result = pd.read_sql_query(query, conn)

result.head(10)
```

```python
query = """
SELECT 
    CustomerID,
    COUNT(*) AS Number_of_Orders,
    SUM(TotalPrice) AS Total_Order_Value,
    AVG(TotalPrice) AS Avg_Order_Value
FROM orders
GROUP BY CustomerID
ORDER BY Total_Order_Value DESC;
"""

result = pd.read_sql_query(query, conn)

result.head(10)
```

```python
query = """
SELECT 
    OrderID,
    Date,
    Product,
    Quantity,
    UnitPrice,
    TotalPrice,
    PaymentMethod,
    OrderStatus
FROM orders
WHERE CustomerID = 'C38840'
ORDER BY Date;
"""

result = pd.read_sql_query(query, conn)

result
```

#### Finding

**Customer C38840** was the only customer meeting both high value order and repeat customers conditions, with **2 orders and 5,723.23 in recorded order value**.

However, both orders were **Returned or Cancelled**, so this customer should not automatically be classified as a successful retained customer.

```python
query = """
SELECT 
    PaymentMethod,
    COUNT(*) AS Number_of_Orders
FROM orders
GROUP BY PaymentMethod
ORDER BY Number_of_Orders DESC;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT 
    PaymentMethod,
    COUNT(*) AS Number_of_Orders,
    SUM(TotalPrice) AS Total_Order_Value,
    AVG(TotalPrice) AS Avg_Order_Value
FROM orders
GROUP BY PaymentMethod
ORDER BY Total_Order_Value DESC;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT 
    ReferralSource,
    COUNT(*) AS Number_of_Orders
FROM orders
GROUP BY ReferralSource
ORDER BY Number_of_Orders DESC;
"""

result = pd.read_sql_query(query, conn)

result
```

### Finding

**Credit Card** generated the highest recorded order value at **263,847.63**, while **Online** payments had the highest order count at **258**.

```python
query = """
SELECT 
    ReferralSource,
    COUNT(*) AS Number_of_Orders,
    SUM(TotalPrice) AS Total_Order_Value,
    AVG(TotalPrice) AS Avg_Order_Value
FROM orders
GROUP BY ReferralSource
ORDER BY Total_Order_Value DESC;
"""

result = pd.read_sql_query(query, conn)

result
```

### Finding

**Instagram** generated the highest order volume (**259**) and highest recorded order value (**275,285.45**).

```python
query = """
SELECT 
    OrderStatus,
    COUNT(*) AS Number_of_Orders
FROM orders
GROUP BY OrderStatus
ORDER BY Number_of_Orders DESC;
"""

result = pd.read_sql_query(query, conn)

result
```

#### Finding

**Cancelled orders** were the largest status category with **250** orders, followed by **Returned with 247**.

An important analytical caution is that TotalPrice represents recorded order value; the dataset does not provide a separate field confirming realized revenue for each status.

```python
query = """
SELECT 
    COUNT(*) AS Total_Orders,
    SUM(
        CASE 
            WHEN OrderStatus IN ('Cancelled', 'Returned') 
            THEN 1 
            ELSE 0 
        END
    ) AS Cancelled_Returned_Orders
FROM orders;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT 
    ROUND(
        100.0 * SUM(
            CASE 
                WHEN OrderStatus IN ('Cancelled', 'Returned') 
                THEN 1 
                ELSE 0 
            END
        ) / COUNT(*), 
        2
    ) AS Cancelled_Returned_Percentage
FROM orders;
"""

result = pd.read_sql_query(query, conn)

result
```

### Interpretation

**41.42%** of recorded orders were either **cancelled or returned**. This represents a substantial proportion of the dataset and may indicate potential issues related to fulfillment, customer satisfaction, product expectations, or the purchasing process.

Further investigation would be required to identify the underlying causes.

```python
query = """
SELECT 
    MIN(Date) AS First_Order_Date,
    MAX(Date) AS Last_Order_Date
FROM orders;
"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """
SELECT 
    strftime('%Y', Date) AS Year,
    COUNT(*) AS Number_of_Orders,
    SUM(TotalPrice) AS Total_Order_Value,
    AVG(TotalPrice) AS Avg_Order_Value
FROM orders
GROUP BY Year
ORDER BY Year;
"""

result = pd.read_sql_query(query, conn)

result
```

#### Finding

**2023** was the strongest full year, generating **552,643.24** from **510 orders**.

2025 should be interpreted cautiously because the dataset only runs through June 30, 2025.

```python
query = """
SELECT 
    strftime('%Y-%m', Date) AS Month,
    COUNT(*) AS Number_of_Orders,
    SUM(TotalPrice) AS Total_Order_Value,
    AVG(TotalPrice) AS Avg_Order_Value
FROM orders
GROUP BY Month
ORDER BY Month;
"""

result = pd.read_sql_query(query, conn)

result
```

#### Finding

The highest-revenue month was **June 2024**, with recorded order value of **68,068.54**.

```python
query = """
SELECT 
    COUNT(*) AS Total_Orders,
    SUM(TotalPrice) AS Total_Order_Value,
    AVG(TotalPrice) AS Avg_Order_Value,
    MAX(TotalPrice) AS Highest_Order_Value,
    MIN(TotalPrice) AS Lowest_Order_Value
FROM orders;
"""

result = pd.read_sql_query(query, conn)

result
```
